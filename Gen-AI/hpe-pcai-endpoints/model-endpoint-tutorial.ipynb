```python
import os
from chat_hpeai import ChatHPEAI

# User-defined configuration
os.environ["AUTH_TOKEN"] = "<your_bearer_token_here>"
os.environ["MODEL_ENDPOINT"] = "<your_model_endpoint_here>"

# Retrieve environment variables
auth_token = os.getenv("AUTH_TOKEN")
endpoint = os.getenv("MODEL_ENDPOINT")

if not auth_token or not endpoint:
    raise ValueError("Ensure AUTH_TOKEN and MODEL_ENDPOINT are set in environment variables")

# Define headers
headers = {
    "Authorization": f"Bearer {auth_token}",
    "Content-Type": "application/json"
}

# Initialize the ChatHPEAI class
chat_api = ChatHPEAI(endpoint_url=endpoint, headers=headers, verify_ssl=False)

# Example: Basic Chat
print("=== Basic Chat Example ===")
response = chat_api.chat(
    model="meta/llama-3.1-8b-instruct",
    prompt="What is the capital of France?",
    max_tokens=100,
    temperature=0.7
)
print("Response:", response)
```
